CONDICIONALES: Pide una calificacion con input(). Si es de 0 a 5.9 imprime "Reprobado", de 6 a 10 "Aprobado" y si esta fuera de 0 a 10 imprime "Calificacion invalida

```python
#3
calificacion = float(input("Ingresa tu calificacion: "))
if 0 <= calificacion <= 5.9:
    print("Reprobado ):")
elif 6<= calificacion <= 10:
    print("¡¡¡Aprobado!!!")
else:
    print("Calificacion invalida X")
```

CICLO FOR:Imprime la tabla de multiplicar del 7 (del 1 al 10) y, aparte, la suma de todos los numeros pares del 1 al 50

```python
#4
numero = int(input("Ingresa el numero del que deseas conocer su tabla de multiplicar"))
print(f"===TABLA DE MULTIPLICAR DEL NUMERO {numero}===")
for i in range(1,11):
    print(f"{numero} X {i} = {numero * i}")
    
suma = 0
for i in range(0,51,2):
    suma += i
print("\n")
print(f"La suma de los numeros pares del rango de 1 hasta 50 es: {suma}")
```

CICLO WHILE:Pide numeros al usuario hasta que escriba salir.Al terminar, imprime cuantos numeros capturo y su suma

```python
#5
contador = 0
suma = 0
while True:
    numero = int(input("Ingresa un numero "))
    contador += 1
    suma += numero
    seguir = input("Deseas seguir:").lower()
    if seguir == "no":
        break
print(f"Ingresaste la cantidad de {contador} numeros y su suma es de {suma}")
```
