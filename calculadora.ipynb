```python
#Funciones
def calcular_promedio(lista):
    if len(lista) == 0:
        return 0
    return sum(lista) / len(lista)

def celsius_a_fahrenheit(c):
    return c * 9 / 5 + 32

def km_a_millas(km):
    return km * 0.621371

def kg_a_libras(kg):
    return kg * 2.20462
```

```python
print(calcular_promedio([8,9,7]))
print(calcular_promedio([]))
print(celsius_a_fahrenheit(100))
print(km_a_millas(10))
print(kg_a_libras(1))
```
